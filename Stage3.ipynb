## 3.1 Handling Missing Values

Missing values can lead to inaccurate results if not handled properly.
Handling missing values ensures data completeness and prevents errors during analysis.
First, the check is for the total missing values in each column.

            *Syntax:df.isnull().sum()*

Depending on the requirement, missing values can be:

 a.Filled with appropriate values (mean, median, mode, or a fixed value)

             *Syntax: df.fillna(value, inplace=True)*
 
 b.Removed if they are very few or irrelevant
 
             *Syntax: df = df.dropna()*

 In this case we have categorical data as well as numberical data for which the customized syntax is used.

             *Syntax:*
                  *for col in df.select_dtypes(include=['float64','int64']).columns:

```python
df.isnull().sum()
```

```python
for col in df.select_dtypes(include=['float64','int64']).columns:
    df[col] = df[col].fillna(df[col].mean())

for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].fillna(df[col].mode()[0])
```

## 3.3 Correcting Data Types

Sometimes data is stored in an incorrect format, such as numeric values stored as text or date values stored as strings. 
Correct data types are essential for performing mathematical operations, comparisons, and time-based analysis.
We convert such columns to the appropriate data types.

            *Syntax: 
                    df["column_name"] = df["column_name"].astype("int")
                    df["date_column"] = pd.to_datetime(df["date_column"])

Since the datatype of DOJ & DOB was given as object, it is changed to datetime format, and the dependents coloumn is chaged to int which was identified as float in initial checks.

```python
df['DOJ'] = pd.to_datetime(df['DOJ'], errors='coerce',dayfirst=True)
df['DOB'] = pd.to_datetime(df['DOB'], errors='coerce',dayfirst=True)
if 'Dependents' in df.columns:
    df['Dependents'] = pd.to_numeric(df['Dependents'], errors='coerce').fillna(0).astype(int)
```

## 3.4 Creating Derived Columns

Derived columns are new features created from existing data to provide additional analytical insights. Feature creation enhances the dataset by adding meaningful attributes that support deeper analysis.

```python
# 1. Annual CTC (Total Cost to Company)
df["Annual_CTC"] = df["Gross Salary"] + df["PF"] + df["HRA"] + df["DA"]

# 2. Experience Band
df["Experience_Band"] = pd.cut(df["Year of Experience"],
                               bins=[0,5,10,20,30,40],
                               labels=["0-5","6-10","11-20","21-30","31+"])

# 3. Age Group
df["Age_Group"] = pd.cut(df["Age"],
                         bins=[18,25,35,45,55,65],
                         labels=["18-25","26-35","36-45","46-55","56+"])

# 4. Service Category
df["Service_Category"] = pd.cut(df["In Company Years"],
                                bins=[0,2,5,10,20,30],
                                labels=["0-2 yrs","3-5 yrs","6-10 yrs","11-20 yrs","21+ yrs"])

# 5. Salary Band
df["Salary_Band"] = pd.cut(df["Salary"],
                           bins=[0,50000,100000,150000,200000],
                           labels=["Low","Medium","High","Very High"])

# 6. Dependent Status
df["Dependent_Status"] = df["Dependents"].apply(lambda x: "Has Dependents" if x > 0 else "No Dependents")

# 7. Joining Year
df["Joining_Year"] = df["DOJ"].dt.year

# 8. Birth Year
df["Birth_Year"] = df["DOB"].dt.year

# 9. Career Stage
df["Career_Stage"] = df["Age"].apply(lambda x:
                                      "Early Career" if x < 30 else
                                      "Mid Career" if x < 45 else
                                      "Senior Career")
```

## 3.6 Aggregating Data

Aggregation is used to summarize data and extract higher-level insights.This helps in understanding trends, averages, and comparisons across different groups.

            *Syntax:
                    aggregated_df = df.groupby("Group_Column")["Target_Column"].agg("mean")

```python
# Department-wise Headcount
df["Headcount"] = 1
dept_headcount = df.groupby("Department")["Headcount"].sum()

# Department-wise Average Annual CTC
dept_avg_ctc = df.groupby("Department")["Annual_CTC"].mean().round(2)

# Experience Band-wise Headcount
exp_band_headcount = df.groupby("Experience_Band",observed=True)["Headcount"].sum()

# Age Group-wise Headcount
age_group_distribution = df.groupby("Age_Group",observed=True)["Headcount"].sum()

# Service Category-wise Headcount
service_category_distribution = df.groupby("Service_Category",observed=True)["Headcount"].sum()

# Gender Distribution
gender_distribution = df.groupby("Sex")["Headcount"].sum()

# Marital Status Distribution
marital_status_distribution = df.groupby("Marital Status")["Headcount"].sum()

# Department-wise Average Salary
dept_avg_salary = df.groupby("Department")["Salary"].mean().round(2)

# Career Stage-wise Average Annual CTC
career_stage_ctc = df.groupby("Career_Stage")["Annual_CTC"].mean().round(2)

# Experience Band-wise Average Salary
exp_band_salary = df.groupby("Experience_Band",observed=True)["Salary"].mean().round(2)

# Dependent Status-wise Average Annual CTC
dependent_ctc = df.groupby("Dependent_Status")["Annual_CTC"].mean().round(2)

# Joining Year-wise Headcount Trend
joining_trend = df.groupby("Joining_Year")["Headcount"].sum()

# Birth Year-wise Workforce Distribution
birth_year_distribution = df.groupby("Birth_Year")["Headcount"].sum()
```

### Correlation Between Department and Salary

Correlation analysis can only be performed between numeric variables.  
In this dataset, the Department column is a categorical (text) variable containing values such as Sales, Finance, Marketing, HR, etc. Since correlation requires numeric inputs, attempting to compute **correlation between Department and Salary will result in error.**

To analyze how salary varies across departments, we use **groupby and pivot table analysis** instead of correlation.  
This approach helps compare **average, median, and spread of salaries** across different departments, which directly supports the project objective of identifying **departmental variations in pay structure**.

This line chart presents the trend of average salaries across different experience bands, making it easier to observe how compensation progresses as employees gain experience and whether growth is consistent across the organization.

# 5.9 Department vs Experience Band Salary Heatmap (Multivariate)
This heatmap displays the average salary across departments and experience bands in a single matrix, providing a comprehensive view of how experience and departmental affiliation together influence compensation patterns.

```python
salary_pivot = pd.pivot_table(df,
                               values="Salary",
                               index="Department",
                               columns="Experience_Band",
                               aggfunc="mean",
                               observed=False)

plt.figure(figsize=(10,6))
sns.heatmap(salary_pivot, annot=True, fmt=".0f", cmap="YlGnBu")
plt.title("Average Salary by Department and Experience Band")
plt.xlabel("Experience Band")
plt.ylabel("Department")
plt.show()
```

# 5.10 Correlation Heatmap
